```python
import os
import glob
import PIL
from PIL import Image, ImageDraw
PIL.Image.MAX_IMAGE_PIXELS = None
```

```python
# Image path
image_path = "redacted"
```

```python
# List of images
full_path_images = glob.glob(os.path.join(image_path, "*.jpg"))
input_names = [x.removeprefix(image_path) for x in full_path_images]
output_names = ["gridview_" + x for x in input_names]
output_names
```

```python
def apply_grid(image_path, output_path, grid_size=252):
    # Load the image
    with Image.open(image_path) as img:
        draw = ImageDraw.Draw(img)
        w, h = img.size

        # Define the grid color
        grid_color = (0, 255, 0)  # Green

        # Draw vertical lines
        for x in range(0, w, grid_size):
            draw.line([(x, 0), (x, h)], fill=grid_color, width=7)

        # Draw horizontal lines
        for y in range(0, h, grid_size):
            draw.line([(0, y), (w, y)], fill=grid_color, width=7)

        # Save the new image
        img.save(output_path)
```

```python
for k,l in enumerate(full_path_images):
    apply_grid(l, image_path + output_names[k] )
```
